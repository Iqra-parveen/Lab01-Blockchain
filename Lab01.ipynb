```python
# Lab Task 1
# Hash Generation and Verification

import hashlib

# Input data
data = "Hello Blockchain"

# Generate SHA-256 hash
hash_value = hashlib.sha256(data.encode()).hexdigest()

# Output
print("Original Data:", data)
print("SHA-256 Hash:", hash_value)
```

```python
# Lab Task 2
# Effect of Data Modification on Hash

import hashlib

# Original Data
data = "Hello Blockchain"
original_hash = hashlib.sha256(data.encode()).hexdigest()

# Modified Data
modified_data = "Hello Blockchain!"
modified_hash = hashlib.sha256(modified_data.encode()).hexdigest()

# Output
print("Original Data:", data)
print("Original Hash:", original_hash)

print("\nModified Data:", modified_data)
print("Modified Hash:", modified_hash)
```
